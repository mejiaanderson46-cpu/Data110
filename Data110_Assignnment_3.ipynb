# Part 1: Scatter plot

```python
import matplotlib.pyplot as plt

# 1. Given data
study_hours = [1, 2, 2.5, 3, 4, 5, 5.5, 6, 7, 8]
exam_scores = [55, 60, 63, 66, 72, 78, 80, 84, 88, 94]

# 2. Canvas size
plt.figure(figsize=(7, 4.5))

# 3. Create the scatter plot
plt.scatter(
    study_hours,
    exam_scores,
    color="royalblue",
    marker="o",
    s=65,
    edgecolors="black",
)

# 4. Labels and Title
plt.title("Study Time vs. Exam Score", fontsize=13)
plt.xlabel("Study Hours", fontsize=11)
plt.ylabel("Exam Score", fontsize=11)

# 5. Grid for readability
plt.grid(True, linestyle="--", alpha=0.5)

# 6. Show the graph
plt.tight_layout()
plt.show()
```

**Observed Relationship:**
There is a strong positive linear relationship between study hours and exam scores. As study time increases, exam scores consistently increase as well.

# Part 2: Line Plot


```python
# Data
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
city_a = [68, 70, 72, 71, 74, 76, 75]
city_b = [62, 64, 67, 66, 69, 71, 70]

# Create plot
plt.figure(figsize=(8, 4.5))
plt.plot(
    days,
    city_a,
    label="City A",
    color="crimson",
    linestyle="-",
    marker="o",
    linewidth=2,
)
plt.plot(
    days,
    city_b,
    label="City B",
    color="teal",
    linestyle="--",
    marker="s",
    linewidth=2,
)

# Labels, Title, and Legend
plt.title(
    "Temperature Over One Week (City A vs. City B)", fontsize=13, pad=12
)
plt.xlabel("Day of the Week", fontsize=11)
plt.ylabel("Temperature (°F)", fontsize=11)
plt.legend()
plt.grid(True, linestyle=":", alpha=0.6)

plt.tight_layout()
plt.show()
```

**Warmer City**
City A was consistenly warmer than City B throught the entire week, staying around 5°F to 6°F higher on every single day.

# Part 3: Bar Graphs — Crime Data Visualization
## Task 3A: Basic Bar Chart for Crime Rates in 2023

```python
import matplotlib.pyplot as plt

# Offenses and 2023 rates per 100,000 people
# (\n breaks long names onto a second line for readability)
offenses = [
    "Aggravated\nAssault",
    "Burglary",
    "Larceny",
    "Motor Vehicle\nTheft",
    "Murder",
    "Rape",
    "Robbery",
]
rates_2023 = [264.1, 250.7, 1347.2, 318.7, 5.7, 38.0, 66.5]

# Create bar chart
plt.figure(figsize=(10, 5))
plt.bar(
    offenses, rates_2023, color="steelblue", edgecolor="black", width=0.6
)

# Titles and labels
plt.title("Offense Rates per 100,000 People (2023)", fontsize=13, pad=12)
plt.xlabel("Offense Type", fontsize=11)
plt.ylabel("Offenses per 100,000 People", fontsize=11)

# Horizontal gridlines
plt.grid(axis="y", linestyle="--", alpha=0.7)

plt.tight_layout()
plt.show()
```

## Task 3B: Change in Crime Rates from 2022 to 2023

```python
import matplotlib.pyplot as plt

# Offense names and percentage changes (2022–2023)
offenses_raw = [
    "Aggravated Assault",
    "Burglary",
    "Larceny",
    "Motor Vehicle Theft",
    "Murder",
    "Rape",
    "Robbery"
]
pct_change = [-3.3, -8.1, -4.9, 12.0, -12.0, -9.8, -0.8]

# Distinct color for each crime type
colors = [
    "#4e79a7", "#f28e2b", "#e15759", "#76b7b2",
    "#59a14f", "#edc948", "#b07aa1"
]

# Create horizontal bar chart
plt.figure(figsize=(9, 5))
plt.barh(offenses_raw, pct_change, color=colors, edgecolor="black")

# Reference line at zero to distinguish increases from decreases
plt.axvline(0, color="black", linewidth=1.2)

# Titles and labels
plt.title("Change in Offense Rates per 100,000 People (2022–2023)", fontsize=13, pad=12)
plt.xlabel("Percentage Change (%)", fontsize=11)
plt.ylabel("Offense Type", fontsize=11)

# Gridlines along the x-axis
plt.grid(axis="x", linestyle="--", alpha=0.6)

plt.tight_layout()
plt.show()
```

**Short Response:**
I think this graph is more useful than Task 3A because it makes the changes between the different crimes easier to see. In Task 3A, crimes like larceny have much higher numbers, so they make the smaller crimes like murder look almost invisible. By looking at the percentage change instead, it is easier to compare the crimes. For example, murder had the biggest decrease at -12.0%, while motor vehicle theft increased by 12.0%. These changes are not as easy to notice when just looking at the absolute rates.


## Task 3C: Improved Clarity Bar Graph

```python
import matplotlib.pyplot as plt

# 1. Dataset paired with changes
crime_data = [
    ("Aggravated Assault", -3.3),
    ("Burglary", -8.1),
    ("Larceny", -4.9),
    ("Motor Vehicle Theft", 12.0),
    ("Murder", -12.0),
    ("Rape", -9.8),
    ("Robbery", -0.8)
]

# 2. Sort from greatest decrease to greatest increase
crime_data.sort(key=lambda item: item[1])
labels = [item[0] for item in crime_data]
changes = [item[1] for item in crime_data]

# 3. Meaningful colors: Green = Crime Drop, Red = Crime Rise
bar_colors = ["#2ca02c" if val < 0 else "#d62728" for val in changes]

# 4. Plot
plt.figure(figsize=(9, 5.2))
bars = plt.barh(labels, changes, color=bar_colors, edgecolor="none", height=0.6)

# Clear baseline divider at 0%
plt.axvline(0, color="gray", linewidth=1.2)

# Titles and axis limits
plt.title("Direction & Magnitude of Crime Rate Changes (2022–2023)\nFBI Data via Brennan Center",
          fontsize=13, weight="bold", pad=15)
plt.xlabel("Percentage Change (%)", fontsize=11)
plt.xlim(-16, 16)
plt.grid(axis="x", linestyle=":", alpha=0.6)

# 5. Add direct value labels beside each bar
for bar in bars:
    width = bar.get_width()
    ha = 'right' if width < 0 else 'left'
    offset = -0.5 if width < 0 else 0.5
    plt.text(width + offset,
             bar.get_y() + bar.get_height() / 2,
             f"{width:+.1f}%",
             va='center',
             ha=ha,
             fontsize=10,
             weight='semibold')

plt.tight_layout()
plt.show()
```

**Clarity Improvements Made:**
1. **Sorted Categories:** Sorted the offenses from the biggest decrease to the biggest increase, which makes it easier to see the biggest changes right away (Murder dropped the most; Motor Vehicle Theft increased the most).
2. **Semantic Color Coding:** Changed the colors so green shows decreases and red shows increases, making it easier to understand the direction of the changes at a quick glance.
3. **Direct Data Labels:** Added exact percentage labels next to each bar to eliminate reading errors across the grid.

## Task 3D: Multi-Series Grouped Bar Chart

```python
import matplotlib.pyplot as plt
import numpy as np

# 1. Dataset
companies = ["Company A", "Company B", "Company C"]
product1 = [20, 34, 30]
product2 = [25, 32, 34]
product3 = [30, 35, 27]

# 2. X-axis positioning and width setup
x = np.arange(len(companies))  # Base tick coordinates [0, 1, 2]
bar_width = 0.25               # Width allocated to each product bar

plt.figure(figsize=(8.5, 5))

# 3. Plot bars with offsets to place them side by side per company
plt.bar(x - bar_width, product1, width=bar_width, label="Product 1", color="#4c72b0", edgecolor="black")
plt.bar(x,             product2, width=bar_width, label="Product 2", color="#dd8452", edgecolor="black")
plt.bar(x + bar_width, product3, width=bar_width, label="Product 3", color="#55a868", edgecolor="black")

# 4. Labels, Title, and Legend
plt.title("Product Comparison Across Companies", fontsize=13, pad=12)
plt.xlabel("Company", fontsize=11)
plt.ylabel("Units Sold / Metric Value", fontsize=11)
plt.xticks(x, companies)
plt.legend(title="Products", frameon=True)
plt.grid(axis="y", linestyle="--", alpha=0.6)

plt.tight_layout()
plt.show()
```
